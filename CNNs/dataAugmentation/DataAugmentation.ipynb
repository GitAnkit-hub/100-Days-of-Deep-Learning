```python
img = image.load_img('/content/drive/MyDrive/CampusX/100 Days of DL/CNNs/dataAugmentation/Copy of CS24MTECH12019.jpg', target_size = (1000, 1000))
```

```python
datagen = ImageDataGenerator(rotation_range = 40,
                             width_shift_range = 0.2,
                             height_shift_range = 0.2,
                             shear_range = 0.2,
                             zoom_range = 0.2,
                             horizontal_flip = True
                             #vertical_flip = True,
                             #fill_mode = 'nearest'
                             )
```

```python
img = image.img_to_array(img)
```

```python
type(img)
```

```python
img.shape
```

```python
input_batch = img.reshape(1, 1000, 1000, 3)
```

```python
i = 0

for output in datagen.flow(input_batch, batch_size = 1, save_to_dir ='/content/drive/MyDrive/CampusX/100 Days of DL/CNNs/dataAugmentation/AugData' ):

      i = i + 1
      if  i == 10:
        break
```
