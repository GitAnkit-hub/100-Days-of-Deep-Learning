```python
import numpy as np
import keras
from keras import layers
```

```python
inputs = keras.Input(shape = (227, 227, 3))
```

```python
inputs.shape
```

```python
inputs.dtype
```

```python
#Build the model

model = keras.Sequential()
# Add an explicit Input layer to define the input shape for the model
model.add(keras.Input(shape=(227, 227, 3))) # Use the actual input shape (height, width, channels)

#First convolutional Layer
# Remove the input_shape argument from the first Conv2D layer as it's now handled by the Input layer
model.add(layers.Conv2D(filters = 96, kernel_size = (11,11), padding = 'valid', strides = (4,4), activation = 'relu'))
model.add(layers.MaxPooling2D(pool_size = (3,3), strides = (2,2))) # Corrected strides from (2.2) to (2,2)

#Second Convolutional Layer
model.add(layers.Conv2D(filters = 256, kernel_size=(5,5), strides = (1,1), padding = 'same', activation ='relu'))
model.add(layers.MaxPooling2D(pool_size = (3,3), strides = (2,2)))

#third Convolutional Layers
model.add(layers.Conv2D(filters = 384, kernel_size = (3,3), padding = 'same', strides = (1,1), activation = 'relu'))
#No pooling

#4th convolutional layer
model.add(layers.Conv2D(filters = 384, kernel_size = (3,3), padding = 'same', strides = (1,1), activation = 'relu'))

#5th Convolutional layer
model.add(layers.Conv2D(filters = 256, kernel_size = (3,3), padding = 'same', strides =(1,1), activation = 'relu'))
model.add(layers.MaxPooling2D(pool_size = (3,3), strides = (2,2)))

# FCC layers
model.add(layers.Flatten())
model.add(layers.Dense(4096, activation = 'relu'))
model.add(layers.Dense(4096, activation = 'relu'))
model.add(layers.Dense(1000, activation = 'softmax'))
```
