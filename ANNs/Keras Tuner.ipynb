## Define Extended Model Building Function

### Subtask:
Create a new Python function, `build_model_extended`, that encapsulates the model architecture with an expanded hyperparameter search space, including the number of layers, units per layer, activation functions, dropout rates, and the learning rate.


**Reasoning**:
The subtask requires defining an extended model building function that incorporates various hyperparameters for tuning. This function will use Keras Tuner's hyperparameter object to define search spaces for the number of layers, units per layer, activation functions, dropout rates, optimizers, and learning rates.



```python
def build_model_extended(hp):
  model = Sequential()

  # Define the number of hidden layers
  num_layers = hp.Int('num_layers', min_value=1, max_value=10, default=2)

  for i in range(num_layers):
    # Define units for each dense layer
    units = hp.Int('units_' + str(i), min_value=8, max_value=128, step=8, default=32)

    # Define activation function for each dense layer
    activation = hp.Choice('activation_' + str(i), values=['relu', 'tanh', 'sigmoid'], default='relu')

    if i == 0:
      model.add(Dense(units=units, activation=activation, input_dim=8))
    else:
      model.add(Dense(units=units, activation=activation))

    # Add Dropout layer after each hidden layer
    dropout_rate = hp.Float('dropout_' + str(i), min_value=0.0, max_value=0.9, step=0.1, default=0.2)
    model.add(tf.keras.layers.Dropout(dropout_rate))

  # Output layer
  model.add(Dense(1, activation='sigmoid'))

  # Define optimizer
  optimizer_choice = hp.Choice('optimizer', values=['adam', 'sgd', 'rmsprop', 'adadelta'], default='adam')

  # Define learning rate
  learning_rate = hp.Float('learning_rate', min_value=1e-4, max_value=1e-2, sampling='log', default=1e-3)

  if optimizer_choice == 'adam':
    optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)
  elif optimizer_choice == 'sgd':
    optimizer = tf.keras.optimizers.SGD(learning_rate=learning_rate)
  elif optimizer_choice == 'rmsprop':
    optimizer = tf.keras.optimizers.RMSprop(learning_rate=learning_rate)
  elif optimizer_choice == 'adadelta':
    optimizer = tf.keras.optimizers.Adadelta(learning_rate=learning_rate)

  model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])

  return model
```

**Reasoning**:
The next step is to create a Keras Tuner RandomSearch object using the newly defined `build_model_extended` function. This will set up the hyperparameter search with the specified objective and project directory.



```python
tuner_extended = kt.RandomSearch(build_model_extended, objective='val_accuracy', max_trials=5, directory='project01', project_name='AllInThree')
```

**Reasoning**:
The previous warning from Keras is advisory and does not prevent the `tuner_extended` object from being created or functioning. The next logical step in the hyperparameter tuning process is to execute the search method to find the best hyperparameters for the model.



```python
tuner_extended.search(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

**Reasoning**:
After the hyperparameter search is complete, the next step is to retrieve the best hyperparameters found by the tuner to understand the optimal configuration for the model.



```python
best_hps = tuner_extended.get_best_hyperparameters(num_trials=1)[0]
print(f"Best Hyperparameters: {best_hps.values}")
```

**Reasoning**:
With the best hyperparameters identified, the next step is to retrieve the optimal model configuration from the tuner to prepare it for full training.



```python
model = tuner_extended.get_best_models(num_models=1)[0]
```

**Reasoning**:
Now that the optimal model has been retrieved with the best hyperparameters, it needs to be trained for the specified number of epochs (100) using the complete training and validation data, starting from the epoch after the hyperparameter search.



```python
model.fit(X_train, y_train, epochs=100, initial_epoch=5, validation_data=(X_test, y_test))
```

**Reasoning**:
After training the optimal model, the next step is to evaluate its performance on the test set to summarize its accuracy and loss, as requested by the task.



```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```
