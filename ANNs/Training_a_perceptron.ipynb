```python
from sklearn.datasets import make_classification
import numpy as np
X,y = make_classification(n_samples = 100, n_features = 2, n_informative = 1, n_redundant = 0,
                          n_classes = 2,  n_clusters_per_class = 1, random_state = 41, hypercube = False, class_sep = 10)
```

```python
def step(z):
    return 1 if z > 0 else 0
```

```python
# THIS is HOW we we get the values of weights and bias
```

```python
def Perceptron(X,y):
    X = np.insert(X, 0, 1, axis = 1) # added a column at index 0 containing all 1's
    weights = np.ones(X.shape[1])  # numpy array with 3 '1s' # initially random weights are chosen
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(100)
        y_hat = np.dot(X[j], weights)  # dot product between weight vector and X vector => region of the random point according to current line weights
        y_hat = step(y_hat)
        weights = weights + lr*(y[j] - y_hat)*X[j]

    return weights[0], weights[1:]  # first column is BIAS and rest starting from index 1 are weights
        
        
        
```

```python
intercept_, coef_ = Perceptron(X,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0] / coef_[1])
b = -(intercept_ / coef_[1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input, y_input, color = 'red', linewidth = 4)
plt.scatter(X[:, 0], X[:, 1], c = y, cmap = 'winter', s = 100)
plt.ylim(-3,2)
```
